```python
from datascience import *
import numpy as np

%matplotlib inline
import matplotlib.pyplot as plots
plots.style.use('fivethirtyeight')
from scipy import stats
from statsmodels.stats.proportion import proportions_ztest
def ztest_prop( successes, observations, prop, alt='two-sided'):
    z,p=proportions_ztest(successes,observations, value = prop, alternative = alt)
    print(f"Z-statistic: {z:.4f}")
    print(f"P-value: {p:.4f}")
```

# Hypothesis Testing

## Example 1: Video Streaming Accounts

A video streaming company historical data shows that $20\%$ of users cancel their subscription within the first 30 days. To reduce churn, the product team redesigns the onboarding experience. In a random sample of 400 new users who went through the new onboarding, 68 canceled their subscription within 30 days.

### Research Question

Is there sufficient evidence at the 0.05 level to conclude that the new onboarding reduced the 30-day cancellation rate?

## Example 2: Campus Coffee Shop Eco-Friendly Cups

A university sustainability office reports that historically $35\%$ of students bring reusable mugs to campus coffee shops. After a campus-wide sustainability campaign, a student researcher surveys a random sample of 250 students buying coffee and finds that 102 brought reusable mugs.

### Research Question

Does the sample provide significant evidence at the $\alpha = 0.01$ level that the proportion of students using reusable mugs has increased?

## Example 3: Quality Control in Organic Produce Packaging

A national supermarket chain mandates that no more than $5\%$ of packaged organic spinach bags can show signs of spoilage before the expiration date. A quality assurance manager randomly selects 500 bags from across several store locations and discovers 33 bags showing early spoilage.

### Research Question

Question: Test at $\alpha = 0.05$ whether the spoilage rate exceeds the supermarket chain's maximum allowed limit.

## Example 4: Public Approval of Municipal Public Transit

In a city polling benchmark from two years ago, $50\%$ of residents favored allocating additional city funds toward expanding public light-rail lines. Following recent city growth, a local news outlet polls a random sample of 600 city residents and finds that 324 favor the funding expansion.

### Research Question:

Is there statistically significant evidence at the $\alpha = 0.05$ level that public support for transit expansion has changed from the $50\%$ baseline?

## Example 5: Online Retail Free Shipping Threshold

An online retailer knows that $15\%$ of site visitors abandon their shopping carts at checkout due to unexpected shipping costs. The company tests a lower "free shipping" threshold banner at the top of the site. Out of 800 visitors tested with this banner, 96 abandoned their carts.

### Research Question: 

Is there sufficient evidence at the $\alpha = 0.05$ level to conclude that the free shipping banner decreased cart abandonment?

## Example 6: College Graduate Job Placement Rate

A career services department advertises that $88\%$ of its business school graduates land a job or enter graduate school within 6 months of graduation. An independent auditing committee selects a random sample of 150 recent graduates and finds that 126 are employed or in graduate school.

### Research Question

Does this sample suggest that the actual placement rate is significantly lower than advertised? Test at $\alpha = 0.05$.

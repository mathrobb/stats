# <span style="font-weight:bold;font-size:1.1em;color:blue;">Analyzing One Categorical Variable</span>

In this notebook we will learn:

- How to analyze a single (binary) categorical variable using the **z-proportion test**.
- About and how to use the 1-sample z-proportion test.

```python
from datascience import *
import numpy as np

%matplotlib inline
import matplotlib.pyplot as plots
plots.style.use('fivethirtyeight')
from scipy import stats
from statsmodels.stats.proportion import proportions_ztest
def ztest_prop( successes, observations, prop, alt='two-sided'):
    z,p=proportions_ztest(successes,observations, value = prop, alternative = alt)
    print(f"Z-statistic: {z:.4f}")
    print(f"P-value: {p:.4f}")
```

## <span style="font-weight:bold;font-size:1.1em;color:blue;">Data Checks for Proportion Testing</span>

We are analyzing a category variable, not a numeric one. Thus, we need to verify that we have **at least 10 items in each sub-category**. This will ensure that we have a large enough sample size for the test to produce accurate results. If sample size is $n$ and the proportion being tested is $p_0$, the formulas are as follows:

- $np_0 \geq 10$
- $n(1-p_0) \geq 10$

## <span style="font-weight:bold;font-size:1.1em;color:blue;">The Z-Proportion Test</span>

- Number of successes
- Number of trials
- Proportion: $p_0$
- alt = (alternative hypothesis)

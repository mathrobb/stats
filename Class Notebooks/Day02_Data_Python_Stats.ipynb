## <span style="color:blue; font-weight:bold; font-size:1.1em">3. Arrays and Arithmetic</span>

Scientific computing: Python can quickly find statistics for arrays:

- mean

- standard deviation
- 90th percentile

When we work in python, we prefer (often require) that the data are in an array. Let's create several arrays with the function:<center><span style="color:purple; font-weight:bold; font-size:1.2em">make_array()</span></center></br>

```python
x = make_array( 10, 12, 14, 15, 15, 16, 18, 20, 22, 24 )
y = make_array( 2, 3, 4, 5, 5, 6, 8, 12, 25, 40 )
z = make_array( 5, 12, 28, 45, 62, 85, 110, 150, 230, 450 )
```

### The **.sort** Method

Often, to quickly find a maximum or minimum value, we can sort a column using the<center><span style="color:darkblue; font-weight:bold; font-size:1.2em">.sort</span></center></br>method.

The **.sort()** sorts in ascending order (1,2,3,... or a,b,c,...) based upon the column title given inside the paranthesis. We have a second input called<center><span style="color:darkblue; font-weight:bold; font-size:1.2em">descending = True</span></center></br>which we can add to reverse the sort.

```python
cones.sort('Price')
```

```python
cones.sort('Price', descending = True)
```

#### **Shortcut**

We can shorten the second input to <center><span style="color:darkblue; font-weight:bold; font-size:1.2em">True</span></center></br> which works as Python knows that only one option exists for the **.sort** method.

```python
cones.sort('Price', True)
```

# Searching Tables: the **.where** method



To find and display the buildings in the table that are located in Los Angeles, we can use:<center><span style="color:darkblue; font-weight:bold; font-size:1.2em">skyscrapers.where('city','Los Angeles')</span></center></br>
Notice with **.where** we have two inputs:
1. Column title indicating **where** to search, and
2. Exact text to search for.

We can also search for values instead of text, and we have **predicates** that are helpers for performing common searches.

```python
# Performing the example shown above
skyscrapers.where('city','Los Angeles')
```

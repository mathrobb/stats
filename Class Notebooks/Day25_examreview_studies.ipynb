## <span style="color:blue;font-weight:bold;font-size:1.1em">Example 3: The Asch Conformity Experiments (1951)</span>
Core Question: Will an individual conform to a group's obviously incorrect answer when faced with social pressure?

Methodology: Solomon Asch placed 1 real participant in a room with 6–8 actors (confederates) who had secretly agreed on their responses in advance. The task was simple: match a line length to three comparison options. On 12 out of 18 critical trials, all confederates gave the exact same wrong answer.

Key Finding: About 75% of participants conformed to the group's wrong answer at least once, demonstrating that social approval and group consensus can override obvious perceptual reality.

## <span style="color:blue;font-weight:bold;font-size:1.1em">Example 4: The Robbers Cave Experiment (1954)</span>
**Research Question:** How do group identities form, and can intergroup conflict be dissolved once established?

Muzafer Sherif and colleagues took twenty-two 11-year-old boys (all similar in background) to a summer camp at Robbers Cave State Park, Oklahoma. The study had 3 distinct phases:

1. **Ingroup Formation:** Split into two isolated groups (<span style="color:red;font-weight:bold">Eagles</span> and <span style="color:red;font-weight:bold">Rattlers</span>) to bond.
2. **Friction Phase:** Pitted against each other in competitive sports with prizes.
3. **Integration Phase:** Forced to work together on superordinate goals requiring joint cooperation (e.g., fixing the camp's broken water supply system).

### Key Finding: Competition quickly led to intense hostility, name-calling, and cabin raids. 
However, hostility could not be erased simply by bringing the groups together for fun activities; it required shared goals that neither group could accomplish alone.

## <span style="color:blue;font-weight:bold;font-size:1.1em">Example 7: The Oregon Health Insurance Experiment (2008)</span>
Researchers attempted to measure the causal impact of **Medicaid health coverage** on health outcomes and financial stability. 

**The study:** Low-income, uninsured adults in Oregon were randomly selected (by state lottery) into a group permitting them to apply for Medicaid coverage. Those who did not "win" in the lottery remained on the standard waiting list.

Gaining Medicaid significantly increased healthcare utilization (more doctor visits and prescription drugs), virtually eliminated catastrophic out-of-pocket medical debt, and lowered depression rates, but showed no significant changes in physical health metrics like blood pressure or blood sugar.

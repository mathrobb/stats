### <span style="color:blue; font-size:1.1em; font-weight:bold">4a. Importing a Table</span> 

For CSV files in the same folder in which are we working, we import that data into a table using<center><span style="color:darkblue; font-weight:bold; font-size:1.2em">Table.read_table(' ')</span></center></br>
where the file name and extension are in quotes.  For example:
<center><span style="color:darkblue; font-weight:bold; font-size:1.2em">Table.read_table('movies_by_year_with_ticket_price.csv')</span></center></br>will load the next table we need.

```python
movies = Table.read_table('movies_by_year_with_ticket_price.csv')
movies.show(5)
```

### <span style="color:blue; font-size:1.1em; font-weight:bold">4b. Extracting a Column</span> 

To retrieve the contents of a column and place them into an array, we use the<center><span style="color:darkblue; font-weight:bold; font-size:1.2em">.column</span></center></br>
method. Below, we:
1. Retrieve the data.
2. Convert to dollars.
3. Calculate number of tickets sold.

```python
gross_in_millions = movies.column('Total Gross')
gross_in_dollars = gross_in_millions * 1e6
tix_sold = gross_in_dollars / movies.column('Average Ticket Price')
```

### <span style="color:blue; font-size:1.1em; font-weight:bold">4c. Add a New Column to our Table</span> 

```python
movies = movies.with_column('Tickets sold', tix_sold)
movies.show(4)
```

### <span style="color:blue; font-size:1.1em; font-weight:bold">4d. Cool Feature: Number Formatter</span>

### Division of Integers = Float

**Most of the time we divide**, Python converts the result to a *float* even when the resulting fraction is very simple.

```python
20 / 2
```

```python
type( 20 / 2 )
```

### Large Values can be Integers

Python uses **scientific notation** for floats. Compare the two results below.

```python
123456789 ** 20
```

```python
123456789.0 ** 20
```

Adding a decimal place (even though it contains $0$) **forces** Python to evaluate the expression as a **float**. This is shown above.

### What Do We Mean by **float**?

**Floating point arithmetic** was developed by computer scientists to add precision to computer calculations otherwise limited by storage (bits and bytes).

```python
# Divsion => floats (even when not needed)
4 / 700
```

```python
# Large numbers + scientific notation means float
4 / 700000000000000000
```

### Limits of Storage Capacity

The number $0.12345678901234567890123456788$ is too large to fit in the usual storage space allocated for a variable. This **leads to an error** when we subtract a similar value.

```python
0.12345678901234567890123456789
```

```python
0.12345678901234567890123456788 - 0.1234567890123456789
```
